```python
!pip install geopandas
import geopandas as gpd
import pyogrio

path = "../data/raw/gadm41_NGA.gpkg"

# See which layers the file contains (we want the LGA level)
print(pyogrio.list_layers(path))
```

```python
lga = gpd.read_file(path, layer="ADM_ADM_2")   # use the layer name you saw above

# Keep only Lagos State
lagos = lga[lga["NAME_1"] == "Lagos"].copy()

print(len(lagos))                 # should be 20
print(lagos["NAME_2"].tolist())   # the LGA names

lagos.plot(edgecolor="black", figsize=(8, 8))
```

```python
print(lagos.columns.tolist())
print(len(lagos))
```

```python
import os

os.makedirs("../data/clean", exist_ok=True)
lagos.to_file("../data/clean/lagos_lgas.gpkg", driver="GPKG")

print(os.path.exists("../data/clean/lagos_lgas.gpkg"))
```

```python
lga = gpd.read_file(path, layer="ADM_ADM_2")   # use your layer name
lagos = lga[lga["NAME_1"] == "Lagos"].copy()
lagos = lagos[["NAME_2", "geometry"]].rename(columns={"NAME_2": "lga"})
```

```python
print(lagos.columns.tolist())   # should show ['lga', 'geometry']
print(len(lagos))               # should show 20

lagos.to_file("../data/clean/lagos_lgas.gpkg", driver="GPKG")

# Reload it to confirm what was saved
check = gpd.read_file("../data/clean/lagos_lgas.gpkg")
print(check.columns.tolist())
print(check["lga"].tolist())
```

```python
check.plot(edgecolor="black", figsize=(8, 8))
```
